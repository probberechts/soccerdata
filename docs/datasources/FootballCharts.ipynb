## Other leagues

Football Charts covers more than 90 leagues, including many lower divisions. To use one that is not in the default league dictionary, add it to `SOCCERDATA_DIR/config/league_dict.json` with the Football Charts identifier (see [Adding custom leagues](../howto/custom-leagues.rst)). For example:

```json
{
  "ENG-Championship": {
    "FootballCharts": "cha",
    "season_start": "Aug",
    "season_end": "May"
  },
  "SWE-Allsvenskan": {
    "FootballCharts": "sweden1",
    "season_start": "Mar",
    "season_end": "Nov"
  }
}
```

The list of identifiers is available at https://footballcharts-backend.onrender.com/api/v1/leagues/.
